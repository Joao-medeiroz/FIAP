```python


dit = {}
dit['Paginas_lidas'] = int(input("Digite o total de paginas lidas: "))
dit['Total_Paginas'] = int(input("Digite o total de paginas: "))

valor = dit['Total_Paginas'] * (dit['Paginas_lidas'] / 100)

if valor >= 60:
    print("Você leu 3/5 do livro")

elif valor >= 80:
    print("Você leu 4/5 do livro")

else:
    print("Leitura insignificante")
```

```python
alunos = {
    "Ana": {"n1": 8, "n2": 7},
    "Bruno": {"n1": 5, "n2": 6}
}

for nome, notas in alunos.items():
    for prova, valor in notas.items():
        print(nome, prova, valor)
```

```python
carrinho = {}
produto = ""

while produto != "sair":
    produto = input("Produto: ")
    if produto == "sair":
        break
    preco = float(input("Preço: "))
    carrinho[produto] = preco

print(carrinho)
```
