```python
class Aluno:
    def __init__(self, nome, nota):
        self.nome = nome
        self.nota = nota

nomes = ["Ana", "Bruna", "Carla", "Diego", "Elisa", "Fabio", "Gina", "Hugo"]
notas = [7, 5, 9, 4, 8, 6, 10 , 3]

alunos =  []
for i in range(8):
    alunos.append(Aluno(nomes[i], notas[i]))

aprovados = 0
for aluno in alunos:
    if aluno.nota >= 6:
        aprovados += 1 

print(aprovados)
```

```python
class Produto:
    def __init__(self, nome, preco):
        self.nome = nome
        self.preco = preco


produtos = []
for i in range(8):
    produtos.append(Produto(f"Item{i+1}", 10 * (i+1)))

for p in produtos:
    print(p.nome)
    
```

```python
class Funcionario: 
    def __init__(self, nome, departamento):
        self.nome = nome
        self.departamento = departamento

funcionarios = [
    Funcionario("Ana", "TI"),
    Funcionario("Bruna", "RH"),
    Funcionario("Carla", "TI"),
    Funcionario("Diego", "Financeiro"),
    Funcionario("Elisa", "RH"),
    Funcionario("Fabio", "TI"),
    Funcionario("Gina", "Financeiro"),
    Funcionario("Hugo", "TI")

]

contagem = {}
for f in funcionarios:
    if f.departamento in contagem:
        contagem [f.departamento] += 1
    else:
        contagem[f.departamento] = 1

print(contagem["TI"])
```
