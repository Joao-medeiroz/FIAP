```python
# Recursão = Resolve problemas fazendo com que uma função chame a si mesma

def fatorial_rec(n):
    if n == 0:
        return 1
    return n * fatorial_rec(n - 1)

%time fatorial_rec(10)
```

```python
# Repetição

def fatorial_it(n):
    resultado = 1
    for i in range (1, n+1):
        resultado *= i

    return resultado
    
%time fatorial_it(10)
```

```python
# Variáveis de entrada

d = [[],[],[]]
c = 0

# Loop
while True:
    # Verificação
    try: 
        # Loop para inserção de 2 valores
        while c < 2:
            print(f"Dados da {c + 1}° pessoa:")
            # Estrutura de Entrada
            n = input("Digite o nome: ")
            i = int(input("Digite a idade: "))
            s = input("Digite M para Masculino ou F para Feminino: ")
            # Adicionando valores na lista
            d[0].append(n)
            d[1].append(n)
            d[2].append(s)
            # Adicionando valor ao contador
            c += 1
        # Estrutura de saída
        print(d)
        # Fim do loop
        break
    # Verifica caso o valor digitado seja incorreto
    except ValueError:
        # Estrutura de saída
        print("Por favor, verifique novamente o valor que você digitou")
```

```python
# Variáveis de entrada

d = [[],[],[]]
c = 0

# Loop
while True:
    # Verificação
    try:
        # Loop para inserção de 2 valores
        while c < 3:
            print(f"Dados da {c + 1}° pessoa:")
            # Estrutura de Entrada
            n = input("Digite o nome: ")
            i = int(input("Digite a idade: "))
            s = input("Digite M para Masculino ou F para Feminino: ")
            # Adicionando valores na lista
            d[c].append(n)
            d[c].append(i)
            d[c].append(s)
            # Adicionando valor ao contador
            c += 1
        # Estrutura de saída
        print("\nLISTA")
        for i in range(3):
            print(f"{i+1}° pessoa: \n{d[i]}")
        # Quebra do Loop
        break
    except ValueError:
        print("Por favor, Verifique o valor que você digitou novamente")
   
```
