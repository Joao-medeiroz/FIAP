```python
import pandas as pd

class Pessoa:
    def __init__(self, nome, salario, taxaDesconto=27.5):
        self.nome = nome
        self.salario = salario
        self.taxaDesconto = taxaDesconto

    def Calculo(self):
        self.novoValor = self.salario - (self.salario * self.taxaDesconto) / 100
        return self.novoValor

    def apresentar(self):
        return f"Olá {self.nome}! Seu salário é de {self.salario}. \nO desconto é de: {self.taxaDesconto}% \nSeu novo salário é: {self.Calculo()}"


metodo = input("Escolha o método: (A/B) ").lower()

if metodo == "a":

    lista = []
    lista1 = []

    for i in range(2):

        valor = Pessoa(
            input("Digite o seu nome: "),
            float(input("Digite o seu salário: "))
        )

        valor.Calculo()

        lista.append(valor.apresentar())

        lista1.append({
            'Nome': valor.nome,
            'Salário': valor.salario,
            'Desconto': valor.taxaDesconto,
            'ValorFinal': valor.novoValor
        })

    for i in range(2):
        print(lista[i])

    dataframe = pd.DataFrame(lista1)
    dataframe.to_excel('dados.xlsx')
    dataframe.to_csv('dados.csv')
    print(dataframe)


else:

    valor = Pessoa(
        input("Digite o seu nome: "),
        float(input("Digite o seu salário: "))
    )

    valor.Calculo()

    dados = [{
        'Nome': valor.nome,
        'Salário': valor.salario,
        'Desconto': valor.taxaDesconto,
        'ValorFinal': valor.novoValor
    }]

    print(pd.DataFrame(dados))
```

```python
from matplotlib import pyplot as plt
import pandas as pd

df1 = pd.read_csv("manutencao_preditiva(in).csv")

x = df1['UDI']
y1 = df1['Temperatura Ar [K]']
y2 = df1['Temperatura Processo [K]']

plt.figure(figsize=(10, 6))

plt.plot(x, y1, label='Temperatura do Ar [K]', color='blue')
plt.plot(x, y2, label='Temperatura do Processo [K]', color='red')

plt.title('Temperatura do Ar e do Processo')
plt.xlabel('UDI')
plt.ylabel('Temperatura [K]')

plt.legend()
plt.grid(False)
plt.show()
```
