```python
class Pilha:
    def __init__(self):
        self.itens = [] # Cria a pilha vazia
    def push(self, item):
        self.itens.append(item) # Adiciona no topo
    def pop(self):
        return self.itens.pop() # Remove do topo
    def show(self):
        return print(self.itens)

# Uso
p = Pilha()
for i in range(1, 101):
    p.push(i)
p.show()
```

```python
class NomeDaClasse:
    def __init__(self, atributo1, atributo2):
        self.atributo1 = atributo1
        self.atributo2 = atributo2
    
    def metodo(self):
        print("Executando método")
        print(self.atributo1)
        print(self.atributo2)

a = NomeDaClasse("A", "B")
a.metodo()
```

```python
class Pessoa:
    def __init__(self, nome, idade):
        self.nome = nome
        self.idade = idade
        
    def get(self):
        return f"Olá sou {self.nome}, tenho {self.idade} anos"

p1 = Pessoa("Arthur", 18)
p2 = Pessoa("Anna", 29)

print(p1.get())
print(p2.get())
```

```python
# Ex. 1
# Crie um classe chamada Estudante, a qual apresente quatro variáveis:
# Nome, faculdade, curso, semestre.
# Após isso, apresente um método para utilizar e mostrar quatro
# variáveis e mostrar um texto usando todas vairáveis.

class Estudante:
    
    def __init__(self, n, f, c, s):
        self.n = n
        self.f = f
        self.c = c
        self.s = s

    def get(self):
        return f"Estudante: {self.n} \nFaculdade: {self.f} \nCurso: {self.c} \nSemestre: {self.s}"

e = Estudante("Anna", "FIAP", "IA", "Terceiro")

print(e.get())
```

```python
class Retangulo: 
    
    def __init__(self, base, altura):
        self.base = base
        self.altura = altura
    
    def area(self):
        return self.base * self.altura
    
    def perimetro(self):
        return 2 * (self.base + self.altura)

r1 = Retangulo(float(input("Base: ")), float(input("Altura: ")))
print(f"A Área do r1: {r1.area()} - Perímetro {r1.perimetro()}")
```

```python
# Ex. 2
# Crie um algoritmo utilizando class para calcular o indice de massa
# corporal para 3 pessoas. Variáveis de entrada: nome, peso e altura
# Cálculo: peso / (altura**2)
# Criar 3 objetos (nome, peso, altura)
# Testar para os 3 objetos

class Massa:

    def __init__(self, n, p, a):
        self.n = n
        self.p = p
        self.a = a

    def set_massa(self):
        return f"Olá {self.n}, a sua massa é de {round(self.p / (self.a ** 2), 2)}"

m = Massa(input("Digite o seu nome: "), float(input("Digite o seu peso")), float(input("Digite a sua altura")))
print(m.set_massa())
```

```python
Olá Anna, a sua massa é de 29.94
# Ex. 3
# Calcular juros compostos de 3 investimentos (CDB, Tesouro Selic, Fundo Y)
# Tendo as seguintes variáveis: nome, capital, taxa (em % ao mês)
# e tempo (meses)
# montante = capital * (1 + taxa / 100) ** tempo
# lucro = montante - capital
# Criar 3 objetos

class Juros:
    def __init__(self, n, c, ta, te):
        self.n = n
        self.c = c
        self.ta = ta
        self.te = te
        self.m = self.c * (1 + self.ta / 100) ** self.te
    
    def montante(self):
        return f"Nome: {self.n} \nMontante: {round(self.m, 2)}"

    def lucro(self):
        return f"Nome: {self.n} \nLucro: {round(self.m - self.c, 2)}"

j = Juros("Anna", 100, 0.10, 3)
%time print(j.montante())
%time print(j.lucro())
```

```python
# Ex. 4
# Crie a classe Livro para catalogar o acervo da biblioteca da FIAP
# Cada livro deve saber montar sua própria ficha catalográfico
# Condição: leitura_longa se paginas > 300
# Saída Esperada
# "Dom Casmurro", de Machado de Assis (256 p.)
# "Guerra e Paz", de Liev Tolstói (1200 p.)
# "O Pequeno Príncipe", de Saint-Exupéry (96 p.)

class Livro:

    def __init__(self, n, a, np):
        self.n = n
        self.a = a
        self.np = np

    def nps(self):
        return f"""Nome: {self.n} 
                   Autor: {self.a} 
                   Número de páginas: {self.np}
                   {"Livro Longo" if self.np > 300 else "Livro Curto"}"""

l1 = Livro("Dom Casmurro", "Machado de Assis", 256)
l2 = Livro("Guerra e Paz", "Liev Tolstói", 1200)
l3 = Livro("O Pequeno Príncipe", "Saint-Exupéry", 96)

print(l1.nps())
print(l2.nps())
print(l3.nps())
```

```python
import pandas as pd

class Massa:

    def __init__(self, n, p, a):
        self.n = n
        self.p = p
        self.a = a

    def set_massa(self):
        return f"Olá {self.n}, a sua massa é de {round(self.p / (self.a ** 2), 2)}"

p1 = Massa("Anna", 90, 1.70)
p2 = Massa("Paulo", 70, 1.69)
p3 = Massa("Carlos", 100, 2.10)

lp = [p1, p2, p3]

dados = [
    
    {
        "nome": p.n,
        "peso": p.p,
        "altura": p.a
    }
    for p in lp
]

df = pd.DataFrame(dados)
print(df)
```
